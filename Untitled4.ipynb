```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

# 1. LOAD DATA (Ensure path matches your Drive or Colab Uploads)
path = '/content/drive/MyDrive/Company.xlsx'
df = pd.read_excel(path, nrows=50)

# 2. CLEANING (Handling strings like "k", "Public", and "Lakh")
# Cleans review_count (e.g., "59.9k Review Public" -> 59.9)
# 2. CLEANING (Safe Version)
# Removes 'k', 'Review', 'Public', and extra spaces
df['review_count_clean'] = (
    df['review_count']
    .astype(str)
    .str.replace('k', '', case=False)
    .str.replace(',', '')
    .str.extract(r'(\d+\.?\d*)')[0] # Extracts only the number part
)

# Convert to float and fill any errors with the median
df['review_count_clean'] = pd.to_numeric(df['review_count_clean'], errors='coerce')
df['review_count_clean'] = df['review_count_clean'].fillna(df['review_count_clean'].median())

# Ensures ratings are numbers
df['ratings'] = pd.to_numeric(df['ratings'], errors='coerce')
df['ratings'] = df['ratings'].fillna(df['ratings'].median())

print("Cleaning complete without errors!")

# --- TASK 1: HEADQUARTERS ---
# --- Task 1: Company Names and Headquarters ---
print("--- Task 1: Top 10 Companies and their Headquarters ---")

# We select the 'name' and 'hq' columns for the first 10 rows
# then use .reset_index(drop=True) to make the table clean (starting from 0)
hq_table = df[['name', 'hq']].head(10)

# Displaying the table nicely
from google.colab import data_table
# This makes the table interactive and neatly formatted in Colab
display(data_table.DataTable(hq_table, include_index=False, num_rows_per_page=10))

# --- TASK 2: BAR CHART (Ratings) ---
plt.figure(figsize=(10, 5))
plt.bar(df['name'].head(10), df['ratings'].head(10), color='skyblue')
plt.title('Task 2: Top 10 Companies by Rating')
plt.xticks(rotation=45)
plt.ylabel('Rating')
plt.show()

# --- TASK 3: FUNNEL CHART (Reviews) ---
fig = px.funnel(df.nlargest(10, 'review_count_clean'), x='review_count_clean', y='name',
                title='Task 3: Top 10 Companies by Review Count')
fig.show()

# --- TASK 4: LINE CHART (Employees) ---
plt.figure(figsize=(10, 5))
plt.plot(df['name'].head(10), df['employees'].head(10), marker='o', color='green', linewidth=2)
plt.title('Task 4: Companies Employee Count-wise')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# --- TASK 5: PIE CHART (5 Companies) ---
plt.figure(figsize=(6, 6))
plt.pie(df['ratings'].head(5), labels=df['name'].head(5), autopct='%1.1f%%', startangle=140)
plt.title('Task 5: Rating Distribution (First 5 Companies)')
plt.show()
```
